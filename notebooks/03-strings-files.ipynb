<h3 style="color:#0D47A1; margin-top:0;">📚 <b>0. Introduction</b></h3>

<p style="color:#424242; line-height:1.6;">
Strings in Python are not merely standalone objects; rather, they are sequences of characters that represent text data. Similar to numerical data types, which consist of sequences of digits, strings consist of sequences of characters. These sequences are fundamental in Python programming, as they are often used to construct and manipulate more complex data structures like lists, tuples, and dictionaries.
</p>

<p style="color:#424242; line-height:1.6;">
Python treats strings as immutable sequences, meaning that once a string is created, its contents cannot be changed. However, various operations can be performed on strings to generate new strings or extract specific information. These operations include concatenation, slicing, formatting, and various methods for searching and transforming the text. Understanding how to effectively manipulate strings is crucial, as strings are one of the most commonly used data types in Python, especially in tasks involving data processing, file handling, and user input.
</p>

<p style="color:#424242; line-height:1.6;">
Moreover, strings serve as the foundation for more advanced objects and data structures. For instance, strings can be used to represent elements within lists or tuples, and they play a critical role in tasks ranging from basic text processing to complex data analysis. As we delve deeper into Python, we will explore the extensive functionality provided by string operations and how they integrate with other Python data types.
</p>

<p style="color:#424242; line-height:1.6;">
In addition to handling strings, Python provides robust support for working with files, which is essential for tasks like reading input data, storing output, or simply managing data over the course of a program's execution. Python uses the <code>open()</code> function to handle files, allowing you to open, read, write, and close them efficiently. 
</p>

<p style="color:#424242; line-height:1.6;">
The <code>open()</code> function is versatile, enabling you to open a file in various modes, such as reading ('r'), writing ('w'), appending ('a'), or even reading and writing in binary mode ('rb', 'wb'). Once a file is opened, you can perform operations like reading the entire file content into a string, reading it line by line, or writing data back into the file. After completing these operations, it is important to close the file using the <code>close()</code> method to free up system resources and ensure data integrity.
</p>

<p style="color:#424242; line-height:1.6;">
Understanding how to work with files in Python is critical, especially when dealing with real-world data that often needs to be read from or written to external files. This capability, combined with string manipulation techniques, forms the backbone of many Python programs, enabling efficient data processing, storage, and retrieval.
</p>

<h4 style="color:#0D47A1; margin-top:0;">🔤 1.1. Indexing strings and slicing</h4>

<p style="color:#424242; line-height:1.6;">
A string is a sequence of ordered and indexed values. This means you can access each element in the sequence by specifying its index, similar to lists. Portions of a string can also be extracted using slices, with the general notation <code>[start=0]:[stop=len][:step=1]</code>. Indexing in Python starts at 0: the 1st element of a list is at index 0, the 2nd is at index 1, and so on. Thus, the n elements of a list are indexed from <i>0</i> to <i>n-1</i>. The examples below illustrate this.
</p>

<p style="color:#424242; line-height:1.6;">
<strong>Indexing:</strong>
</p>
<p style="color:#424242; line-height:1.6;">
To access a specific character in a string, you use the index inside square brackets. For example:
</p>
<ul style="color:#424242; line-height:1.7;">
  <li><code>ch[0]</code> returns the first character of the string <code>ch</code>.</li>
  <li><code>ch[2]</code> returns the third character.</li>
  <li><code>ch[-1]</code> returns the last character of the string.</li>
</ul>

<p style="color:#424242; line-height:1.6;">
<strong>Slicing:</strong>
</p>
<p style="color:#424242; line-height:1.6;">
Slicing allows you to extract a portion of the string. The syntax for slicing is <code>[start:stop:step]</code>, where:
</p>
<ul style="color:#424242; line-height:1.7;">
  <li><code>start</code> is the index of the first character you want to include in the slice.</li>
  <li><code>stop</code> is the index of the character just after the last character you want to include.</li>
  <li><code>step</code> specifies the interval between characters in the slice (optional; default is 1).</li>
</ul>

<p style="color:#424242; line-height:1.6;">
Examples of slicing:
</p>
<ul style="color:#424242; line-height:1.7;">
  <li><code>ch[:6]</code> returns a substring from the beginning of <code>ch</code> up to, but not including, index 6.</li>
  <li><code>ch[6:]</code> returns a substring from index 6 to the end of the string.</li>
  <li><code>ch[0:10:2]</code> returns every second character from index 0 to 10.</li>
</ul>

<p style="color:#424242; line-height:1.6;">
<strong>Examples:</strong>
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
ch = "Christelle"<br>
print(ch[0])    # Returns 'C'<br>
print(ch[2])    # Returns 'r'<br>
print(ch[-1])   # Returns 'e', the last character<br>
print(ch[:6])   # Returns 'Christ'<br>
print(ch[6:])   # Returns 'elle'<br>
print(ch[0:10:2]) # Returns 'Crsel'
</div>

<p style="color:#424242; line-height:1.6;">
These operations make it easy to manipulate and extract data from strings, which is crucial for text processing and data manipulation in Python.
</p>

</div>

```python
ch="Christelle"
print(ch[0]) # returns 'C'
print(ch[2]) # returns 'r'
print(ch[-1]) # returns 'e', the last element of ch
print(ch[:6]) # returns 'Christ'
print(ch[6 :]) # returns 'elle'
print(ch[0:10:2]) # returns 'Crsel'
```

<h4 style="color:#388E3C; margin-top:0;">🔗 1.3. Addition of strings (Concatenation)</h4>

<p style="color:#424242; line-height:1.6;">
String addition refers to the process of placing strings together, one after the other, to form a single string. This operation is known as <b>concatenation</b>. To perform concatenation, the <code>+</code> symbol is used between the string variable names. The resulting string is then assigned to a new variable. Examples:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
str1 = "Hello"<br>
str2 = "World"<br>
result = str1 + " " + str2  # Concatenates 'Hello' and 'World' with a space in between<br>
print(result)  # Outputs: Hello World
</div>

</div>

```python
x = "One small step for man,"
y = "one giant leap for mankind."
z = x + " " + y  # Concatenates the two strings with a space in between
z  # Outputs: 'Un petit pas pour l'homme, un grand pas pour l'humanité'
```

<h4 style="color:#0D47A1; margin-top:0;">⚠️ Concatenation with Numeric Values</h4>

<p style="color:#424242; line-height:1.6;">
Be cautious when using the concatenation operator to combine a numeric value with a string to form a single string. For example:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
x = 'The price of the pen is'<br>
y = 5<br>
z = 'dollars'
</div>

<p style="color:#424242; line-height:1.6;">
If you want to concatenate <code>x</code>, <code>y</code>, and <code>z</code> to obtain the sentence 'The price of the pen is 5 dollars', you can't directly do <code>x + y + z</code> because <code>y</code> is a numeric type. You need to first convert <code>y</code> to a string using the <code>str()</code> function. Here's how it should be done:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
result = x + " " + str(y) + " " + z<br>
print(result)  # Outputs: 'The price of the pen is 5 dollars'
</div>

</div>

```python
x = 'The price of the pen is'
y = 5
z = 'dollars'
ch = x + " " + " " + str(y) + " " + z
ch # returns 'The price of the pen is5euros'
```

<h4 style="color:#388E3C; margin-top:0;">🔧 1.4. Methods</h4>

<p style="color:#424242; line-height:1.6;">
Strings come with <a href="https://docs.python.org/2/library/stdtypes.html#string-methods" target="_blank" style="color:#1565C0; text-decoration:none;">numerous features</a>—referred to as <code>methods</code> in Object-Oriented Programming (OOP)—that facilitate their manipulation. The <code>dir()</code> function lists all available methods for strings. Among these, the most commonly used are <code>upper()</code>, <code>lower()</code>, <code>capitalize()</code>, <code>find()</code>, <code>replace()</code>, <code>count()</code>, <code>startswith()</code>, <code>endswith()</code>, and the <code>in</code> keyword. 
</p>

<p style="color:#424242; line-height:1.6;">
These methods allow for operations such as converting text to different cases, searching for and replacing substrings, and checking string content for specific patterns. They are crucial for text processing and data manipulation tasks, enabling efficient handling and transformation of string data. Understanding and utilizing these methods effectively can greatly enhance the capability to work with and analyze text-based information in Python.
</p>

</div>

```python
title = "Python course"
help(title.replace)
```

```python
enfant, peluche = "Calvin", 'Hobbes'  # Multiple assignment

titre = enfant + ' et ' + peluche  # +: String concatenation
print(titre)
print("================================================")
print('\t')

print(titre.replace('et', '&'))  # Substring replacement: 'Calvin & Hobbes'
print("================================================")
print('\t')

print(' & '.join(titre.split(' et ')))  # Splitting and joining
print("================================================")
print('\t')

print('Hobbes' in titre)  # in: Inclusion test
print("================================================")
print('\t')

print(titre.find("Hobbes"))  # str.find: Substring search

print(titre.center(30, '-'))  # '-------Calvin et Hobbes-------'

dir(str)  # Lists all string methods
```

</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">

<p style="color:#424242; line-height:1.6;">
We want to display a message like <code>"The price of the pen is 2 euros"</code>.

<p style="color:#424242; line-height:1.6;">
The goal of formatting is to embed the value of a variable within a string. There are three primary methods to achieve this:
</p>

<ul style="color:#424242; line-height:1.7;">
  <li>Using the concatenation operator <code>+</code></li>
  <li>Using the formatting operator <code>%</code></li>
  <li>Using the <code>str.format()</code> method</li>
</ul>

<p style="color:#424242; line-height:1.6;">
We will illustrate just one example using the <code>str.format()</code> method. The formatting system allows for precise control over the conversion of variables into strings. It primarily relies on the <code>str.format()</code> method, which enables detailed specification of how variables should be inserted and formatted within a string. For more information, you can refer to the <a href="https://docs.python.org/2/library/string.html#formatstrings" target="_blank" style="color:#1565C0; text-decoration:none;">formatting documentation</a>.
</p>

```python
print("{nom} a {age} ans".format(nom='Calvin', age=6)) # 'Calvin a 6 ans'
print("================================================")
print('\t')

print("{} a {} ans".format('Calvin', 6)) # Shortcut 'Calvin a 6 ans'
print("================================================")
print('\t')

pi = 3.1415926535897931
print("{x:f} {x:.2f} {y:f} {y:g}".format(x=pi, y=pi*1e9)) # '3.141593 3.14 3141592653.589793 3.14159e+09'
```

<h3 style="color:#0D47A1; margin-top:0;">📁 2. File I/O</h3>

<p style="color:#424242; line-height:1.6;">
File I/O (Input/Output) in Python involves reading from and writing to files. Python provides several built-in functions and methods to handle files. The primary function for file manipulation is <code>open()</code>, which allows you to open a file and returns a file object. This file object provides methods for reading and writing data. Here's a detailed overview with various examples:
</p>

<h4 style="color:#1976D2; margin-top:20px;">Basic File Operations</h4>

<p style="color:#424242; line-height:1.6;">
<strong>1. Opening and Closing a File</strong><br>
To open a file, use the <code>open()</code> function. The basic syntax is <code>open(filename, mode)</code>, where <code>filename</code> is the name of the file and <code>mode</code> is the mode of operation (<code>'r'</code> for read, <code>'w'</code> for write, <code>'a'</code> for append, etc.). To close a file, use the <code>close()</code> method of the file object.
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
# Opening a file in read mode<br>
file = open('example.txt', 'r')<br>
# Perform operations on the file<br>
file.close()
</div>

<p style="color:#424242; line-height:1.6;">
<strong>2. Reading from a File</strong><br>
<code>read()</code> method reads the entire content of the file. <code>readline()</code> method reads a single line from the file, and <code>readlines()</code> method reads all lines into a list.
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
# Read entire file<br>
file = open('example.txt', 'r')<br>
content = file.read()<br>
print(content)<br>
file.close()<br><br>
# Read line by line<br>
file = open('example.txt', 'r')<br>
line = file.readline()<br>
while line:<br>
&nbsp;&nbsp;&nbsp;&nbsp;print(line, end='')<br>
&nbsp;&nbsp;&nbsp;&nbsp;line = file.readline()<br>
file.close()<br><br>
# Read all lines into a list<br>
file = open('example.txt', 'r')<br>
lines = file.readlines()<br>
print(lines)<br>
file.close()
</div>

<p style="color:#424242; line-height:1.6;">
<strong>3. Writing to a File</strong><br>
<code>write()</code> method writes a string to the file. <code>writelines()</code> method writes a list of strings to the file.
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
# Write a string to a file<br>
file = open('example.txt', 'w')<br>
file.write('Hello, World!')<br>
file.close()<br><br>
# Write multiple lines to a file<br>
lines = ['First line\n', 'Second line\n', 'Third line\n']<br>
file = open('example.txt', 'w')<br>
file.writelines(lines)<br>
file.close()
</div>

<p style="color:#424242; line-height:1.6;">
<strong>4. Appending to a File</strong><br>
To append data to an existing file, open the file in append mode (<code>'a'</code>).
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
file = open('example.txt', 'a')<br>
file.write('This is an appended line.\n')<br>
file.close()
</div>

<p style="color:#424242; line-height:1.6;">
<strong>5. Using <code>with</code> Statement</strong><br>
The <code>with</code> statement is used for resource management. It ensures that the file is properly closed after its suite finishes, even if an exception is raised.
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
# Reading a file using with statement<br>
with open('example.txt', 'r') as file:<br>
&nbsp;&nbsp;&nbsp;&nbsp;content = file.read()<br>
&nbsp;&nbsp;&nbsp;&nbsp;print(content)<br><br>
# Writing to a file using with statement<br>
with open('example.txt', 'w') as file:<br>
&nbsp;&nbsp;&nbsp;&nbsp;file.write('This is a new content.')
</div>

<p style="color:#424242; line-height:1.6;">
<strong>6. File Modes</strong><br>
<code>'r'</code>: Read (default mode, opens the file for reading)<br>
<code>'w'</code>: Write (creates a new file or truncates an existing file)<br>
<code>'a'</code>: Append (opens the file for appending)<br>
<code>'b'</code>: Binary (reads or writes in binary mode, e.g., <code>'rb'</code> or <code>'wb'</code>)<br>
<code>'t'</code>: Text (default mode, opens the file in text mode, e.g., <code>'rt'</code> or <code>'wt'</code>)
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
# Binary mode example<br>
with open('example.jpg', 'rb') as file:<br>
&nbsp;&nbsp;&nbsp;&nbsp;content = file.read()<br>
&nbsp;&nbsp;&nbsp;&nbsp;# Process binary data
</div>

<p style="color:#424242; line-height:1.6;">
<strong>7. File Positioning</strong><br>
<code>seek(offset, whence)</code>: Moves the file pointer to a specific position.<br>
<code>tell()</code>: Returns the current file pointer position.
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
with open('example.txt', 'r') as file:<br>
&nbsp;&nbsp;&nbsp;&nbsp;file.seek(10)  # Move to the 10th byte<br>
&nbsp;&nbsp;&nbsp;&nbsp;print(file.tell())  # Print the current file pointer position<br>
&nbsp;&nbsp;&nbsp;&nbsp;content = file.read(20)  # Read 20 bytes from the current position<br>
&nbsp;&nbsp;&nbsp;&nbsp;print(content)
</div>

<p style="color:#424242; line-height:1.6;">
<strong>8. Working with File Paths</strong><br>
Use the <code>os</code> and <code>pathlib</code> modules to handle file paths.
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">

</div>

<strong>9. Handling File Exceptions</strong><br>
Use <code>try</code> and <code>except</code> blocks to handle file-related exceptions.
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
