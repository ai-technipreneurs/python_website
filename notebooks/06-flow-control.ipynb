<h4 style="color:#388E3C; margin-top:0;">⚖️ 0.0. Instructions with the if...else clause</h4>

<p style="color:#424242; line-height:1.6;">
Conditional statements are operations that the program executes when a logical expression evaluates to <code>True</code> or <code>False</code> (depending on the case). Conditional statements are defined within an <code>if...else</code> clause.
</p>

<p style="color:#424242; line-height:1.6;">
The general structure of an <code>if...else</code> block is as follows:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
if logical_expression:<br>
&nbsp;&nbsp;&nbsp;&nbsp;statement1<br>
&nbsp;&nbsp;&nbsp;&nbsp;statement2<br>
&nbsp;&nbsp;&nbsp;&nbsp;...<br>
&nbsp;&nbsp;&nbsp;&nbsp;statementn<br>
else:<br>
&nbsp;&nbsp;&nbsp;&nbsp;other_block_of_statements
</div>

<p style="color:#424242; line-height:1.6;">
Note the <code>:</code> symbol following the <code>if</code> or <code>else</code> keyword, which indicates the end of the <code>if</code> or <code>else</code> declaration and the beginning of the statement definitions. Also, note that the <code>else</code> clause is optional, meaning there can be an <code>if</code> clause without an <code>else</code> clause.
</p>

<p style="color:#424242; line-height:1.6;">
It is crucial to observe the essential role of <i>indentation</i>, which delineates each block of statements, and the presence of colons after the condition and after the <code>else</code> keyword.
</p>

</div>

````python
# Instructions with a simple "if" clause
a = 150
if (a > 100):
    print("a exceeds 100")

print("================================================")
print('\t')


# Instructions with the if...else clause
x = 65
if (x > 70):
    print("x exceeds 70")
else:
    print("x does not exceed 100")
````

<h4 style="color:#0D47A1; margin-top:0;">🔄 0.2. Defining nested <code>if</code> clauses</h4>

<p style="color:#424242; line-height:1.6;">
The previous examples illustrate first-level <code>if</code> clauses. In complex programs, it is very common to have nested <code>if</code> clauses, meaning that <code>if</code> clauses are defined inside other <code>if</code> clauses, sometimes at multiple levels. See the example below:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
a = 15<br>
b = 10<br><br>
if a > b:<br>
&nbsp;&nbsp;&nbsp;&nbsp;print("a is greater than b")<br>
&nbsp;&nbsp;&nbsp;&nbsp;if a - b > 5:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;print("The difference between a and b is greater than 5")<br>
&nbsp;&nbsp;&nbsp;&nbsp;else:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;print("The difference between a and b is 5 or less")<br>
else:<br>
&nbsp;&nbsp;&nbsp;&nbsp;print("a is not greater than b")
</div>

<p style="color:#424242; line-height:1.6;">
In this example, there is an <code>if</code> statement nested within another <code>if</code> statement. This structure allows for more granular control over the conditions and the corresponding actions in your program.
</p>

</div>

````python
a = 15
b = 10

if a > b:
    print("a is greater than b")
    if a - b > 5:
        print("The difference between a and b is greater than 5")
    else:
        print("The difference between a and b is 5 or less")
else:
    print("a is not greater than b")
````

````python
# Nested if clauses for animal classification
phylum = "vertebrates"
animal_class = "mammals"
order = "carnivores"
family = "felines"

if phylum == "vertebrates":
    if animal_class == "mammals":
        if order == "carnivores":
            if family == "felines":
                print("It might be a cat")
            else:
                print("It's a carnivore but not a feline")
        else:
            print("It's a mammal but not a carnivore")
    else:
        print("It's a vertebrate but not a mammal")
else:
    print("Not a vertebrate")
````

<h3 style="color:#0D47A1; margin-top:0;">🔄 1. Loop Instructions</h3>

<p style="color:#424242; line-height:1.6;">
In Python, there are two main types of loop instructions:
</p>

<ol style="color:#424242; line-height:1.7;">
  <li><strong><code>while...</code> loops</strong>: These loops execute a block of code repeatedly as long as a specified condition remains <code>True</code>. They are useful when the number of iterations is not known in advance and depends on dynamic conditions evaluated during execution.</li>
  <li><strong><code>for... in...</code> loops</strong>: These loops iterate over a sequence (such as a list, tuple, or string) or other iterable objects. They are ideal for iterating through known sequences or ranges of values, allowing for clear and concise iteration over collections or ranges.</li>
</ol>

<p style="color:#424242; line-height:1.6;">
Both types of loops are fundamental for controlling the flow of execution and repeating tasks efficiently in Python programs.
</p>

<h4 style="color:#388E3C; margin-top:0;">⚡ 1.0. while... loops</h4>

<p style="color:#424242; line-height:1.6;">
After introducing conditional structures, we turn our attention to loop structures: these structures allow a block of instructions to be executed multiple times. These repetitions fall into two categories:
</p>

<ul style="color:#424242; line-height:1.7;">
  <li><strong>Conditional repetitions</strong>: The block of instructions is repeated as long as a condition is true.</li>
  <li><strong>Unconditional repetitions</strong>: The block of instructions is repeated a specified number of times.</li>
</ul>

<p style="color:#424242; line-height:1.6;">
The general syntax for defining <code>while</code> loop instructions is as follows:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
Initialize increment variable<br>
while condition:<br>
&nbsp;&nbsp;&nbsp;&nbsp;block_of_instructions<br>
&nbsp;&nbsp;&nbsp;&nbsp;increment
</div>

<p style="color:#424242; line-height:1.6;">
As with the <code>if</code> structure, the condition is first evaluated, and if it is true, the block of instructions is executed. However, with the <code>while</code> loop, after executing the block of instructions, the condition is evaluated again. This process repeats until the condition becomes false. Therefore, it is essential to define an increment variable whose value changes after each execution so that the condition can eventually become false. This increment is necessary to prevent the instructions from being evaluated indefinitely, creating an infinite loop (or endless loop). Such a situation requires forcibly stopping the program's execution.
</p>

</div>

````python
x = 1 # Initialization of the variable x
while (x < 10):
    print('The value of x is ', x)
    x = x + 1 # Increment of the variable x
````

````python
fruits = ['apples', 'oranges', 'strawberries', 'bananas']
i = 0
while i < len(fruits):
    print (fruits[i])
    i = i + 1
````

<h4 style="color:#0D47A1; margin-top:0;">🔄 1.1. Loop instructions: <code>for... in...</code></h4>

<p style="color:#424242; line-height:1.6;">
When we want to repeat a block of instructions a specified number of times, we can use a <i>counter</i>, which is a variable that counts the number of repetitions and controls the exit of the <code>while</code> loop. This is illustrated in the following example, where a function takes an integer <code>n</code> as an argument and prints the same message <code>n</code> times.
</p>

<p style="color:#424242; line-height:1.6;">
To perform such a repetition, we have a loop structure that saves us from initializing the counter <code>(i = 0)</code> and incrementing it <code>(i += 1)</code>: this is the structure introduced by the <code>for</code> keyword.
</p>

<p style="color:#424242; line-height:1.6;">
The general syntax for defining <code>for... in...</code> loop instructions is as follows:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
for element in sequence_of_values :<br>
&nbsp;&nbsp;&nbsp;&nbsp;block of instructions
</div>

</div>

````python
n = 5
for i in range(n):
    print('I repeat myself {} times.' ' (i={})'.format(n, i))
````

````python
# Simple "for...in..." loop
for i in range(1,11) :
    print(i)
````

````python
# "for...in..." loop iterating over a string
listch = "Hello world"
for i in listch :
    print ( i )
````

````python
# Combining a `for...in...` Loop with an `if` Clause
listnb = [4, 5, 6]
for i in listnb:
    if i == 5:
        print("The condition is verified for the element", i)
    else :
        print("The condition is not verified for the element", i)
````

````python
# Combining a `for...in...` Loop with an `if` Clause
mych = "Hello World"
for lettre in mych :
    if lettre in "AEIOUYaeiouy":
        print ('The letter', lettre, 'is a vowel')
    else :
        print ('The letter', lettre, 'is a consonant')
````

````python
# Combining a `for...in...` Loop with an `if` Clause
# Considering the `space` character
mych = "Hello World"
for letter in mych:
    if letter in "AEIOUYaeiouy":
        print('The letter', letter, 'is a vowel')
    elif letter == " ":
        print("This is likely a space")
    else:
        print('The letter', letter, 'is a consonant')
````

````python
fruits = ['mango', 'orange', 'apple', 'banana']
costs = [49, 99, 15, 32]
for fruit, price in zip(fruits, costs):
    print("A", fruit, "costs", price, "cents.")
````

````python
costs = {'mango': 49, 'orange': 99, 'apple': 15, 'banana': 32}
for fruit, price in costs.items():
    print("A", fruit, "costs", price, "cents.")
````

<h3 style="color:#0D47A1; margin-top:0;">✨ 2. List Comprehension</h3>

<p style="color:#424242; line-height:1.6;">
In Python, a convenient and expressive syntax for creating lists is provided by list comprehensions. This method allows you to generate lists in a very concise way, often eliminating the need for explicit loops when elements need to be tested or processed before being included in the list.
</p>

<p style="color:#424242; line-height:1.6;">
The syntax for defining a list comprehension is similar to the mathematical notation for defining a set comprehension:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
[expression for item in iterable if condition]
</div>

<p style="color:#424242; line-height:1.6;">
Here's a breakdown of the components:
</p>

<ul style="color:#424242; line-height:1.7;">
  <li><strong><code>expression</code></strong>: The value or transformation to apply to each item.</li>
  <li><strong><code>item</code></strong>: The variable representing each element in the iterable.</li>
  <li><strong><code>iterable</code></strong>: The collection or sequence you are iterating over.</li>
  <li><strong><code>condition</code></strong> (optional): A filter that determines which items to include.</li>
</ul>

<p style="color:#424242; line-height:1.6;">
<strong>Example:</strong>
</p>

<p style="color:#424242; line-height:1.6;">
Suppose you want to create a list of squares for all even numbers between 1 and 10. Using a list comprehension, you can achieve this concisely:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
squares = [x**2 for x in range(1, 11) if x % 2 == 0]<br>
print(squares)
</div>

<p style="color:#424242; line-height:1.6;">
<strong>Explanation:</strong>
</p>

<ul style="color:#424242; line-height:1.7;">
  <li><code>x**2</code> is the expression that computes the square of each item.</li>
  <li><code>x</code> is the item variable iterating over the numbers in <code>range(1, 11)</code>.</li>
  <li><code>range(1, 11)</code> is the iterable providing numbers from 1 to 10.</li>
  <li><code>if x % 2 == 0</code> is the condition filtering only even numbers.</li>
</ul>

<p style="color:#424242; line-height:1.6;">
The output will be:
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
[4, 16, 36, 64, 100]
</div>

<p style="color:#424242; line-height:1.6;">
This approach is both readable and efficient for generating lists based on specific criteria or transformations.
</p>

</div>

````python
liste = [2, 4, 6, 8, 10]

print([3*x for x in liste])
print("================================================")
print('\t')

print([[x, x**3] for x in liste])
print("================================================")
print('\t')

print([3*x for x in liste if x > 5])  # filtering with a condition
print("================================================")
print('\t')

print([3*x for x in liste if x**2 < 50])  # same as above
print("================================================")
print('\t')

liste2 = list(range(3))
print([x*y for x in liste for y in liste2])
````

````python
# Here is an example of an efficient way to get a list of leap years within a given range:
leap_years = [year for year in range(2000, 2100) if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0)]

print(leap_years)

<div style="text-align: right; margin-top: 12px; font-size: 0.85em;"><a href="#table-of-contents" style="color: #667eea; text-decoration: none;">↑ Back to TOC</a></div>
````

<h3 style="color:#0D47A1; margin-top:0;">⏩ 3. Break and Continue Statements</h3>

<p style="color:#424242; line-height:1.6;">
The reserved keywords <code>break</code> and <code>continue</code> are used to alter the behavior of <code>for...in</code> or <code>while...</code> loops. The <code>break</code> statement allows for the termination of the loop and the exit from it prematurely, even if the main condition defining the loop remains true. On the other hand, the <code>continue</code> statement is used to skip the execution of the remaining instructions in the current iteration of the loop and proceed to the next iteration when the main condition is satisfied. The following examples illustrate their use.
</p>

</div>

```python
# Loop with the break statement
for i in range(5):
    if i > 2:
        break
print(i)
```

````python
# Loop through the sequence
for i in range(5):
    if i > 2:
        break
print(i)
````

box-shadow: 0 2px 6px rgba(0,0,0,0.05);
font-family:'Segoe UI', sans-serif;
">

<h4 style="color:#388E3C; margin-top:0;">📋 Explanation</h4>

<ul style="color:#424242; line-height:1.7;">
  <li>The loop iterates over a range of values from 0 to 4.</li>
  <li>When <code>i</code> becomes greater than 2, the <code>break</code> statement is executed, which exits the loop immediately.</li>
  <li>Therefore, the loop stops, and the value of <code>i</code> at the time the loop is exited is printed. In this case, the output will be <code>3</code>.</li>
</ul>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0; line-height: 1.5;">
# Loop with the continue statement<br>
for i in range(5):<br>
&nbsp;&nbsp;&nbsp;&nbsp;if i == 2:<br>
&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;continue<br>
&nbsp;&nbsp;&nbsp;&nbsp;print(i)
</div>

````python
# Loop through the sequence
for i in range(5):
    if i == 2:
        continue
    print(i)
````
