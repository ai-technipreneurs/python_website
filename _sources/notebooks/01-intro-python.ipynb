</p>

<h4 style="color:#388E3C; margin-top:0;">✂️ <b>List Slicing</b></h4>

<p style="color:#424242; line-height:1.6;">
The expression <code>liste_B[0:2]</code> is used to create a sublist from <code>liste_B</code>, including elements from index <code>0</code> up to but not including index <code>2</code>. In Python, slicing is performed using the format <code>list[start:end]</code>, where <code>start</code> is the index to begin the slice (inclusive) and <code>end</code> is the index to end the slice (exclusive).
</p>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0;">
# Given list_B
liste_B = [0, 3, 209, 4025, 554, 6, 1]

# Creating a sublist from index 0 to 2 (not including 2)
sous_liste = liste_B[0:2]
print(sous_liste)  # Output: [0, 3]
</div>

<p style="color:#424242; line-height:1.6;">
Here, <code>sous_liste</code> will contain the elements <code>[0, 3]</code>, which are the elements at indices <code>0</code> and <code>1</code> of <code>liste_B</code>.
</p>

</div>

```python
liste_B[0:2] #  Sublist, runs throuh liste_B at the indices 0 and 1.
```

<h4 style="color:#0D47A1; margin-top:0;">🎯 <b>Slicing with Step</b></h4>

<p style="color:#424242; line-height:1.6;">
The expression <code>liste_B[0:5:2]</code> is used for slicing with a step. It extracts elements from <code>liste_B</code> starting at index <code>0</code>, up to but not including index <code>5</code>, with a step of <code>2</code>.
</p>

<p style="color:#424242; line-height:1.6;">
In this case:
</p>
<ul style="color:#424242; line-height:1.6;">
  <li><code>0</code> is the starting index (inclusive).</li>
  <li><code>5</code> is the ending index (exclusive).</li>
  <li><code>2</code> is the step, meaning every second element is selected.</li>
</ul>

<div style="background: #2E3440; color: #D8DEE9; padding: 12px; border-radius: 6px; font-family: 'Monaco', 'Consolas', monospace; margin: 10px 0;">
# Given list_B
liste_B = [0, 3, 209, 4025, 554, 6, 1]

# Slicing with start=0, end=5, and step=2
sous_liste = liste_B[0:5:2]
print(sous_liste)  # Output: [0, 209, 554]
</div>

<p style="color:#424242; line-height:1.6;">
In this example, <code>sous_liste</code> contains <code>[0, 209, 554]</code>, which are the elements at indices <code>0</code>, <code>2</code>, and <code>4</code> of <code>liste_B</code>.
</p>

</div>

```python
liste_B[0:5:2] # start:end:step
```

## Programming Structures

- Blocks are defined by **indentation** (usually by 4 spaces);
- One statement per line *generally* (or statements separated by `;`);
- **Comments** start with `#` and extend to the end of the line;
- **Boolean Expression**: a condition is an expression that evaluates to `True` or `False`:
  - `False`: false logical test (e.g., 3 == 4), null value, empty string (''), empty list ([]), etc.,
  - `True`: true logical test (e.g., 2 + 2 == 4), any non-null value or object (and thus evaluating to True by default *except exceptions*);
  - **Logical Tests**: `==`, `!=`, `>`, `>=`, etc.;
  - **Logical Operators**: `and`, `or`, `not`;
  - **Ternary Operator**: `value **if** condition **else** value`;
- **Conditional Expression**: `**if** condition1 : ... [**elif** condition2 : ...] [**else**: ...]`;  
- **For Loop**: `**for** element **in** iterable`, executes on each element of an iterable object:
  - `continue`: interrupts the current iteration and resumes the loop at the next iteration,
  - `break`: completely interrupts the loop;
- **While Loop**: `while condition`: repeats as long as the condition is true, or after an explicit exit with `break`.

These structures will be discussed in detail in upcoming notebooks.
